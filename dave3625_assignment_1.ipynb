```python
# Task 2
# Are there any null values in the Grc_df? If yes, then in which columns and how many? 
# Finally, handle these null values using any strategy shown during the labs.

Grc_df = pd.read_csv("grocery_dataset.csv")
print("\nColumns with null before being handled: ") 
print(Grc_df.isnull().sum()) # Shows that columns Item_Weight and Outlet_Size have null

Grc_df.dropna(inplace=True) # Drops all rows that are null

print("\nColumns with null after being handled: ") 
print(Grc_df.isnull().sum()) # Shows that all nulls have been handled
```

```python
# Task 3
# How many unique Outlet Sizes are there in the Grc_df? Which outlet size is maximum, and which is minimum?

print("Amount of unique outlets: ", Grc_df["Outlet_Size"].nunique()) # show how many unique outlets we have which is 3.
print("Which outlet is maximum? Answer: ", Grc_df["Outlet_Size"].max())
print("Which outlet is minimum? Answer: ", Grc_df["Outlet_Size"].min())

```

```python
#Task 4 
#How many unique Item Fat Content types are in the Grc_df? List them. Do you see any
#issues with the Item Fat Content types? If yes, then handle this issue.

print("Amount of unique Item Fat Content types: ", Grc_df["Item_Fat_Content"].nunique()) 
print("Unique Item Fat Content types: ", Grc_df["Item_Fat_Content"].unique())

Grc_df["Item_Fat_Content"] = Grc_df["Item_Fat_Content"].replace({
    "LF": "Low Fat",
      "low fat": "Low Fat", 
      "reg": "Regular"
        })

print("After handling the issue:")
print(Grc_df["Item_Fat_Content"].unique()) 
print("Number of unique types:", Grc_df["Item_Fat_Content"].nunique())



```

```python
#Task 5
#Drop the columns having index values of 0, 6 and create a new DataFrame Grc_new_df.

Grc_new_df = Grc_df.drop(Grc_df.columns[[0, 6]], axis=1)

display(Grc_new_df.head()) 

print(Grc_new_df.columns)
```

```python
#Task 6
#Using different Supermarket type listed in the column Outlet_Type create two different
#DataFrames from Grc_new_df. Name these DataFrames as SupType_1 and SupType_2.

print(Grc_new_df["Outlet_Type"].unique())

SupType_1 = Grc_new_df[Grc_new_df["Outlet_Type"] == "Supermarket Type1"]
SupType_2 = Grc_new_df[Grc_new_df["Outlet_Type"] == "Supermarket Type2"]

display(SupType_1.head())
display(SupType_2.head())
```

```python
#Task 7 SupType_1 has the higher median MRP (144.6102 vs. 130.1810).
# There are no outliers visible in either box plot.
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

fig, axes = plt.subplots(2, 1, figsize=(8, 10))
sns.boxplot(y=SupType_1["Item_MRP"], ax=axes[0])
sns.boxplot(y=SupType_2["Item_MRP"], ax=axes[1])
print("Median SupType_1:", SupType_1["Item_MRP"].median())
print("Median SupType_2:", SupType_2["Item_MRP"].median())
```

```python
# Task 8 The Outlet_Location_Type of the store with the lowest sales is Tier 3.

Grc_Concat_df = pd.concat([SupType_1, SupType_2])
Grc_Concat_df.head()
Grc_Concat_df = Grc_Concat_df.sort_values(by="Item_Outlet_Sales", ascending=True)
Grc_Concat_df.head()
Grc_Concat_df.iloc[0]["Outlet_Location_Type"]
```

```python
#Task 9 (double index

Grc_Concat_df = Grc_Concat_df.set_index(
    ["Outlet_Size", "Outlet_Location_Type"]
)

Grc_Concat_df.head()
```

```python
# Task 10, cut into 10 buckets and compute mean, min, max + count

# splitting into buckets
Grc_Concat_df["Item_Weight_Bucket"] = pd.cut(
    Grc_Concat_df["Item_Weight"],
    bins=10)

# Calculating those stats
Weight_Summary = (Grc_Concat_df.groupby(
    "Item_Weight_Bucket", observed=False)["Item_Weight"]
.agg(["mean", "min", "max", "count"])
)

Weight_Summary.round(2) # 2 decimal places just easier to read
```
